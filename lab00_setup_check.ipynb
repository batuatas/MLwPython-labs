```python
URL = "https://raw.githubusercontent.com/batuatas/MLwPython-labs/main/data/student_survey.csv"
df = pd.read_csv(URL)
df.head()
```

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.scatter(df["study_hours"], df["grade"], alpha=0.6)
ax.set_xlabel("study hours per week")
ax.set_ylabel("grade (1-10)")
ax.set_title("Your setup works")
plt.show()
```
